```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 60)
print("DATA CLEANING & REPORTING AUTOMATION")
print("=" * 60)

data = {
    "Customer_ID": [
        "C001", "C002", "C003", "C004", "C005",
        "C006", "C007", "C008", "C009", "C010",
        "C010", "C011", "C012", "C013", "C014"
    ],

    "Name": [
        "Rahul", "Priya", "Arun", "Sneha", "Kiran",
        "Anita", "Ravi", "Meena", "Vijay", "Pooja",
        "Pooja", "Suresh", "Divya", "Manoj", "Kavya"
    ],

    "Age": [
        22, 28, 35, np.nan, 42,
        31, 25, 29, 38, 27,
        27, 45, np.nan, 33, 26
    ],

    "City": [
        "Bangalore", "bangalore", "Mysore", "Mysuru", "Bangalore",
        "Chennai", "chennai", "Bangalore", "Mysore", "Bangalore",
        "Bangalore", "Chennai", "Mysore", "Bangalore", "Bangalore"
    ],

    "Purchase_Amount": [
        2500, 4500, 3200, 5000, np.nan,
        2800, 3500, 6200, 4100, 3000,
        3000, 7200, 3800, 4500, 2700
    ],

    "Status": [
        "Completed", "completed", "Pending", "Completed", "pending",
        "Completed", "COMPLETED", "Pending", "completed", "Cancelled",
        "Cancelled", "Completed", "pending", "Completed", "completed"
    ]
}

df = pd.DataFrame(data)

print("\nORIGINAL DATA")
print("-" * 60)
display(df)

print("\nDATASET INFORMATION")
print("-" * 60)

print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])

print("\nColumn Data Types:")
print(df.dtypes)


# ------------------------------------------------------------
# 3. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMISSING VALUES BEFORE CLEANING")
print("-" * 60)

missing_values = df.isnull().sum()
print(missing_values)


df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill Purchase Amount with median
df["Purchase_Amount"] = df["Purchase_Amount"].fillna(
    df["Purchase_Amount"].median()
)

print("\nMissing values handled successfully.")

before_duplicates = len(df)

df = df.drop_duplicates()

after_duplicates = len(df)

duplicates_removed = before_duplicates - after_duplicates

print("\nDUPLICATE RECORD CLEANING")
print("-" * 60)

print("Records before removing duplicates :", before_duplicates)
print("Records after removing duplicates  :", after_duplicates)
print("Duplicates removed                 :", duplicates_removed)

# Standardize City names
df["City"] = df["City"].str.strip().str.title()

# Standardize Status values
df["Status"] = df["Status"].str.strip().str.title()

print("\nINCONSISTENT DATA CLEANED")
print("-" * 60)

print("City values:")
print(df["City"].unique())

print("\nStatus values:")
print(df["Status"].unique())

print("\nDATA VALIDATION")
print("-" * 60)

print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate records:")
print(df.duplicated().sum())

print("\nAge range:")
print(df["Age"].min(), "to", df["Age"].max())

print("\nPurchase amount range:")
print(
    df["Purchase_Amount"].min(),
    "to",
    df["Purchase_Amount"].max()
)

print("\nCLEANED DATA")
print("-" * 60)

display(df)

print("\nSTATISTICAL SUMMARY")
print("-" * 60)

summary = df.describe()

display(summary)

total_customers = df["Customer_ID"].nunique()
total_purchase = df["Purchase_Amount"].sum()
average_purchase = df["Purchase_Amount"].mean()
maximum_purchase = df["Purchase_Amount"].max()
minimum_purchase = df["Purchase_Amount"].min()

report = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Total Purchase Amount",
        "Average Purchase Amount",
        "Maximum Purchase",
        "Minimum Purchase",
        "Duplicates Removed"
    ],

    "Value": [
        total_customers,
        total_purchase,
        average_purchase,
        maximum_purchase,
        minimum_purchase,
        duplicates_removed
    ]
})

print("\nAUTOMATED REPORT")
print("-" * 60)

display(report)

city_report = df.groupby("City").agg(
    Customers=("Customer_ID", "count"),
    Total_Purchase=("Purchase_Amount", "sum"),
    Average_Purchase=("Purchase_Amount", "mean")
).reset_index()

print("\nCITY-WISE REPORT")
print("-" * 60)

display(city_report)

status_report = df.groupby("Status").agg(
    Number_of_Customers=("Customer_ID", "count"),
    Total_Amount=("Purchase_Amount", "sum")
).reset_index()

print("\nSTATUS-WISE REPORT")
print("-" * 60)

display(status_report)

plt.figure(figsize=(9, 5))

plt.bar(
    city_report["City"],
    city_report["Total_Purchase"]
)

plt.title("City-wise Total Purchase")
plt.xlabel("City")
plt.ylabel("Total Purchase Amount")
plt.xticks(rotation=30)
plt.grid(axis="y")

plt.show()

plt.figure(figsize=(8, 5))

plt.bar(
    status_report["Status"],
    status_report["Number_of_Customers"]
)

plt.title("Customer Status Distribution")
plt.xlabel("Status")
plt.ylabel("Number of Customers")
plt.grid(axis="y")

plt.show()

plt.figure(figsize=(9, 5))

plt.hist(
    df["Purchase_Amount"],
    bins=6,
    edgecolor="black"
)

plt.title("Purchase Amount Distribution")
plt.xlabel("Purchase Amount")
plt.ylabel("Number of Customers")
plt.grid(True)

plt.show()


highest_city = city_report.loc[
    city_report["Total_Purchase"].idxmax(),
    "City"
]

highest_status = status_report.loc[
    status_report["Number_of_Customers"].idxmax(),
    "Status"
]

print("\nKEY BUSINESS INSIGHTS")
print("=" * 60)

print("Total Customers:", total_customers)

print(
    "Total Purchase Amount: ₹",
    f"{total_purchase:,.2f}"
)

print(
    "Average Purchase Amount: ₹",
    f"{average_purchase:,.2f}"
)

print(
    "City with highest purchase:",
    highest_city
)

print(
    "Most common customer status:",
    highest_status
)

print(
    "Duplicate records removed:",
    duplicates_removed
)

df.to_csv(
    "cleaned_customer_data.csv",
    index=False
)

print("\nCleaned CSV file created:")
print("cleaned_customer_data.csv")

with pd.ExcelWriter(
    "automated_data_cleaning_report.xlsx",
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned Data",
        index=False
    )

    report.to_excel(
        writer,
        sheet_name="Summary Report",
        index=False
    )

    city_report.to_excel(
        writer,
        sheet_name="City Report",
        index=False
    )

    status_report.to_excel(
        writer,
        sheet_name="Status Report",
        index=False
    )

    summary.to_excel(
        writer,
        sheet_name="Statistics"
    )

print("\nExcel report created:")
print("automated_data_cleaning_report.xlsx")

print("\n" + "=" * 60)
print("DATA CLEANING & REPORTING AUTOMATION COMPLETED")
print("=" * 60)

print("""
The project successfully performed:

1. Data loading
2. Missing value detection
3. Missing value handling
4. Duplicate removal
5. Inconsistent data cleaning
6. Data validation
7. Statistical analysis
8. Automated reporting
9. Data visualization
10. CSV export
11. Excel report generation
""")
```
